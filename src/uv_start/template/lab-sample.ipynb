```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from colors import COLOR

plt.style.use("hhlab_style01.mplstyle")
```

## Some example data

Replace this with your own: put files in `data/` and read them with `pd.read_csv("data/your_file.csv")`.

```python
rng = np.random.default_rng(0)
df = pd.DataFrame(
    {
        "dose_uM": np.repeat([0, 1, 5, 10], 10),
        "viability": np.concatenate(
            [rng.normal(mean, 4, 10) for mean in [98, 90, 70, 45]]
        ),
    }
)
df.groupby("dose_uM")["viability"].agg(["mean", "std"]).round(1)
```

```python
fig, ax = plt.subplots()
ax.scatter(df["dose_uM"], df["viability"], color=COLOR.BLUE.value)
ax.set_xlabel("dose [µM]")
ax.set_ylabel("viability [%]")
plt.tight_layout()

# Figures go in results/, raw data stays untouched in data/
fig.savefig("results/viability.png")
```
